The structure is:
Container
  Layout Row (wrapping Left Column and Meter Bar)
    Content Column (Left Column)
      Title
      Main Content Row (wrapping Image)
      Description Row
    Meter Bar (Right Column)

So:
1. `layout-row` needs to `flex: 1` to fill the container vertically.
2. `meter-bar` needs to stretch to bottom (it’s a child of layout-row, so `align-items: stretch` on layout-row is needed, which is default or needs explicit set).
3. `content-column` needs `flex: 1` (or 3 as currently set) to fill width, but also vertical height?
    - `layout-row` is flex-row. `content-column` will stretch in height.
4. Inside `content-column` (which is flex-col):
    - `title-objective`
    - `main-content-row`
    - `description-row`
    - To "fill in the space" and "make image section larger":
      - `main-content-row` (wrapping image) should have `flex: 1` to consume available vertical space.
      - `image-interactive` inside it should grow.
